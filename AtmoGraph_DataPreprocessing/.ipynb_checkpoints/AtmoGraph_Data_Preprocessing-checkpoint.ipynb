```python
# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate percentage of missing values in each column

missing_percentage = (df.isnull().mean() * 100).round(2)

print("Missing value percentage:")
print(missing_percentage)
```

```python
# Total missing values in the complete dataset

total_missing = df.isnull().sum().sum()

print("Total missing values in dataset:", total_missing)
```

```python
# Count completely duplicated rows

duplicate_rows = df.duplicated().sum()

print("Total duplicate rows:", duplicate_rows)
```

```python
# Check duplicate disruption IDs

duplicate_ids = df["disruption_id"].duplicated().sum()

print("Duplicate disruption IDs:", duplicate_ids)
```

```python
# Count unique disruption IDs

unique_ids = df["disruption_id"].nunique()

print("Total unique disruption IDs:", unique_ids)
print("Total dataset rows:", len(df))
```

```python
# Keep a copy of the original data before cleaning

df_raw = df.copy()

print("Original dataset backup created!")
```

```python
# Identify text columns

text_columns = df.select_dtypes(include=["object", "string"]).columns

print("Text columns:")
print(text_columns.tolist())
```

```python
# Remove extra spaces at the beginning and end of text values

for col in text_columns:
    df[col] = df[col].str.strip()

print("Extra spaces removed from text columns!")
```

```python
# Display unique categories in categorical columns

for col in text_columns:
    if col != "disruption_id":
        print("\nColumn:", col)
        print(df[col].unique())
```

```python
# Statistical summary of numerical columns

df.describe()
```

```python
# Validate numerical values against expected ranges

validation_checks = {
    "supplier_tier": df["supplier_tier"].between(1, 4),
    "disruption_severity": df["disruption_severity"].between(1, 5),
    "production_impact_pct": df["production_impact_pct"].between(0, 100),
    "revenue_loss_usd": df["revenue_loss_usd"] >= 0,
    "response_time_days": df["response_time_days"] >= 0,
    "partial_recovery_days": df["partial_recovery_days"] >= 0,
    "full_recovery_days": df["full_recovery_days"] >= 0
}

for column, check in validation_checks.items():
    invalid_count = (~check).sum()
    print(f"{column}: {invalid_count} invalid values")
```

```python
# Check whether partial recovery takes longer than full recovery

recovery_errors = df[
    df["partial_recovery_days"] > df["full_recovery_days"]
]

print("Recovery time inconsistencies:", len(recovery_errors))

# Display suspicious records
display(recovery_errors.head(10))
```

```python
# Check Boolean values

print("Backup supplier values:")
print(df["has_backup_supplier"].value_counts())

print("\nPermanent supplier change values:")
print(df["permanent_supplier_change"].value_counts())
```

```python
# Detect outliers in revenue loss using IQR method

Q1 = df["revenue_loss_usd"].quantile(0.25)
Q3 = df["revenue_loss_usd"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
revenue_outliers = df[
    (df["revenue_loss_usd"] < lower_bound) |
    (df["revenue_loss_usd"] > upper_bound)
]

print("Total revenue loss outliers:", len(revenue_outliers))
print("Percentage of dataset:", round(len(revenue_outliers) / len(df) * 100, 2), "%")
```

```python
display(revenue_outliers.head(10))
```

```python
# Create a cleaned copy of the dataset

df_clean = df.copy()

# Create an outlier flag for revenue loss

df_clean["revenue_loss_outlier"] = (
    (df_clean["revenue_loss_usd"] < lower_bound) |
    (df_clean["revenue_loss_usd"] > upper_bound)
)

# Count outliers and normal records

print("Outliers:", df_clean["revenue_loss_outlier"].sum())
print("Normal records:", (~df_clean["revenue_loss_outlier"]).sum())
```

```python
# Standardize text values

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

# Standardize categorical column formatting

categorical_columns = [
    "disruption_type",
    "industry",
    "supplier_region",
    "supplier_size",
    "response_type"
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].str.strip()

print("Categorical data standardization completed!")
```

```python
# Final consistency checks

print("Checking final dataset...")

# Check missing values
print("Missing values:", df_clean.isnull().sum().sum())

# Check duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())

# Check duplicate IDs
print("Duplicate IDs:", df_clean["disruption_id"].duplicated().sum())

# Check recovery time consistency
recovery_errors = df_clean[
    df_clean["partial_recovery_days"] >
    df_clean["full_recovery_days"]
]

print("Recovery inconsistencies:", len(recovery_errors))

# Check production impact range
print("Invalid production impact:",
      (~df_clean["production_impact_pct"].between(0, 100)).sum())

# Check revenue loss
print("Negative revenue losses:",
      (df_clean["revenue_loss_usd"] < 0).sum())

print("\nFinal validation completed!")
```

```python
## Linking Disruptions with Supply Chain Nodes
import numpy as np
import pandas as pd

# Select supplier nodes from our nodes dataset
supplier_nodes = nodes_df[nodes_df["node_type"] == "Supplier"].copy()

# Use a fixed seed so the mapping stays reproducible
np.random.seed(42)

# Assign a supplier node to every disruption record
df_linked = df_clean.copy()

df_linked["impacted_node_id"] = np.random.choice(
    supplier_nodes["node_id"],
    size=len(df_linked)
)

# Add node information to each disruption
df_linked = df_linked.merge(
    supplier_nodes[["node_id", "node_name", "node_type", "city", "country"]],
    left_on="impacted_node_id",
    right_on="node_id",
    how="left"
)

# Rename columns for clarity
df_linked.rename(columns={
    "node_name": "impacted_node_name",
    "node_type": "impacted_node_type",
    "city": "impacted_city",
    "country": "impacted_country"
}, inplace=True)

# Remove the extra node_id column after merging
df_linked.drop(columns=["node_id"], inplace=True)

display(df_linked.head())
print("Total disruption records:", len(df_linked))
```

```python
print("Missing node mappings:",
      df_linked["impacted_node_id"].isnull().sum())

print("\nDisruptions assigned to each node:")
print(df_linked["impacted_node_id"].value_counts())

print("\nDataset shape:", df_linked.shape)
```

```python
os.makedirs("data/processed", exist_ok=True)

df_linked.to_csv(
    "data/processed/disruptions_linked_to_nodes.csv",
    index=False
)

print("Linked disruption dataset saved successfully!")
```
