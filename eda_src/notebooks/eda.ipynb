```python
import pandas as pd 

df = pd.read_csv('../../data/supply_chain_disruption_recovery_dataset.csv')

df.head()
```

```python
print(df.dtypes)
```

## Univariat Analysis 

```python
# Summary statistic for  numerical column

numeric_cols = ['supplier_tier','disruption_severity','production_impact_pct','revenue_loss_usd']

summary = df[numeric_cols].describe().T
summary['median'] = df[numeric_cols].median()
summary['IQR'] = summary['75%'] - summary['25%']
summary['skewness'] = df[numeric_cols].skew()
print("Numeric Summary Statistics:\n")

print(summary)
```
