```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns

df = pd.read_csv('../../data/supply_chain_disruption_recovery_dataset.csv')

df.head()
```

```python
print(df.dtypes)
```

## Univariat Analysis 

```python
# Summary statistic for  numerical column

numeric_cols = ['supplier_tier','disruption_severity','production_impact_pct','revenue_loss_usd']

summary = df[numeric_cols].describe().T
summary['median'] = df[numeric_cols].median()
summary['IQR'] = summary['75%'] - summary['25%']
summary['skewness'] = df[numeric_cols].skew()
print("Numeric Summary Statistics:\n")

print(summary)
```

```python
# Histogram, boxplot, density plot for numerical columns

for col in numeric_cols:
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 3, 1)
    sns.histplot(df[col], kde=True)
    plt.title(f'Histogram of {col}')

    plt.subplot(1, 3, 2)
    sns.boxplot(y=df[col])
    plt.title(f'Boxplot of {col}')

    plt.subplot(1, 3, 3)
    sns.kdeplot(df[col], fill=True)
    plt.title(f'Density Plot of {col}')

    plt.tight_layout()
    plt.show()
```

```python
# Frequency count for Categorical columns

cat_cols = ['disruption_type','industry','supplier_region','supplier_size','has_backup_supplier']

for index, col in enumerate(cat_cols):
    print(f"\n Frequency count for {col}:\n", df[col].value_counts())
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=df, order=df[col].value_counts().index)
    plt.title(f'Bar Chart of {col}')
    plt.show()
```
